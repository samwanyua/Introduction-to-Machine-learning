```python
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
sns.set()
plt.quiver(0,0,4,5)
plt.show()
```

```python
plt.quiver(0,0,4,5, scale_units='xy', angles='xy', scale=1, color='b')
plt.quiver(0,0,-4,8, scale_units='xy', angles='xy', scale=1, color='g')
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

Addition of two vectors

```python
vector_1 = np.asarray([0,0,2,3])
vector_2 = np.asarray([0,0,3,-2])
sum = vector_1 + vector_2
sum
```

```python
plt.quiver(0,0,2,3, scale_units='xy', angles='xy', scale=1, color='b')
plt.quiver(0,0,3,-2, scale_units='xy', angles='xy', scale=1, color='g')
plt.quiver(0,0,5,1, scale_units='xy', angles='xy', scale=1, color='r')

plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

Subtracting vectors

```python
vector_1 = np.asarray([0,0,2,3])
vector_2 = np.asarray([0,0,3,-2])
diff = vector_1 - vector_2
diff
```

```python
plt.quiver(0,0,2,3, scale_units='xy', angles='xy', scale=1, color='b')
plt.quiver(0,0,3,-2, scale_units='xy', angles='xy', scale=1, color='g')
plt.quiver(0,0,-1,5, scale_units='xy', angles='xy', scale=1, color='r')

plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

Multiplying a vector by a scalar

```python
vector_3 = np.asarray([0,0,3,1])
vector_4 = 2 * vector_3
vector_4
```

```python
plt.quiver(0,0,3,1, scale_units='xy', angles='xy', scale=1, color='b')
plt.quiver(0,0,6,2, scale_units='xy', angles='xy', scale=1, color='r')

plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```
